```python
from sklearn.model_selection import train_test_split
import numpy as np
import pandas as pd
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
```

```python
class MyMLR:
    def __init__(self):
        self.coef_ = None
        self.intercept_ = None
        
    def fit(self,X_train,Y_train):
        X_train = np.c_[np.ones(X_train.shape[0]), X_train]
        betas = np.dot(np.linalg.inv(np.dot(X_train.T,X_train)) , np.dot(X_train.T,Y_train))
        self.intercept_ = betas[0]
        self.coef_ = betas[1:]
        print(betas)
        
    def predict(self,X_test):
        Y_pred = self.intercept_ + np.dot(X_test, self.coef_)
        return Y_pred
```

```python
diabetes = load_diabetes()
df = pd.DataFrame(data = diabetes.data , columns = diabetes.feature_names)
df.head()
```

```python
X = df
y = diabetes.target
X_train,X_test,Y_train,Y_test = train_test_split(X,y,test_size=0.25,random_state=59)
lr = MyMLR()
lr.fit(X_train,Y_train)
predictions = lr.predict(X_test)
r2_score(Y_test,predictions)
```
